This notebook develops the mathematical foundations for portfolio construction.
I will:
- estimate expected returns for each asset,
- construct the covariance matrix,
- represent portfolio allocations using weights,
- calculate expected portfolio return,
- calculate portfolio variance and volatility,
- examine how diversification affects portfolio risk.

```python
import numpy as np
import pandas as pd
import yfinance as yf
import matplotlib.pyplot as plt
```

```python
tickers = [
    "AAPL",
    "MSFT",
    "JPM",
    "JNJ",
    "XOM",
    "PG",
    "CAT",
    "NEE"
]

start_date = "2016-01-01"
end_date = "2026-01-01"
```

```python
data = yf.download(
    tickers,
    start=start_date,
    end=end_date,
    auto_adjust=True,
    progress=False
)

prices = data["Close"]

returns = prices.pct_change().dropna()
```

## 1. Expected Returns

Let $\mu$ denote the vector of expected annual returns for the assets.

Daily historical mean returns are annualized using approximately 252 trading days per year.

```python
mu = returns.mean() * 252

mu
```

## 2. Covariance Matrix

Portfolio risk depends not only on the volatility of individual assets, but also on how their returns move together.

The covariance matrix captures both effects.

```python
cov_matrix = returns.cov() * 252

cov_matrix
```

## 3. Portfolio Weights

A portfolio can be represented by a vector of weights, where each weight represents the fraction of capital invested in an asset.

For a fully invested long-only portfolio:

$$
\sum_{i=1}^{N} w_i = 1
$$

```python
n_assets = len(tickers)

weights = np.ones(n_assets) / n_assets

weights
```

```python
weights.sum()
```

## 4. Portfolio Expected Return

The expected return of a portfolio is the weighted average of the expected returns of its assets:

$$
E[R_p] = w^\top \mu
$$

For the equal-weight portfolio, the estimated annual expected return is approximately 19.83%.

## 5. Portfolio Risk

Portfolio variance incorporates both individual asset risk and the covariance between asset returns:

$$
\sigma_p^2 = w^\top \Sigma w
$$

Portfolio volatility is the square root of portfolio variance:

$$
\sigma_p = \sqrt{w^\top \Sigma w}
$$

The equal-weight portfolio has an estimated annual volatility of approximately 17.17%.

```python
portfolio_return = weights @ mu
portfolio_return
print(f"Expected annual return: {portfolio_return:.2%}")
```

```python
portfolio_variance = weights @ cov_matrix @ weights
portfolio_variance
```

```python
portfolio_volatility = np.sqrt(portfolio_variance)
print(f"Annual volatility: {portfolio_volatility:.2%}")
```
