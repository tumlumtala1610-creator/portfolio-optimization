# Baseline Portfolio

This notebook constructs an equal-weight portfolio to serve as a benchmark for later portfolio optimization.

The portfolio is evaluated using:
- expected annual return,
- annual volatility,
- Sharpe ratio.

Each asset receives the same portfolio weight.

```python
import numpy as np
import pandas as pd
import yfinance as yf
```

```python
tickers = [
    "AAPL",
    "MSFT",
    "JPM",
    "JNJ",
    "XOM",
    "PG",
    "CAT",
    "NEE"
]

start_date = "2016-01-01"
end_date = "2026-01-01"
```

```python
data = yf.download(
    tickers,
    start=start_date,
    end=end_date,
    auto_adjust=True,
    progress=False
)

prices = data["Close"]

returns = prices.pct_change().dropna()
```

## 1. Return and Risk Estimates

Historical daily returns are used to estimate annual expected returns and the annual covariance matrix.

```python
mu = returns.mean() * 252
cov_matrix = returns.cov() * 252
```

## 2. Equal-Weight Portfolio

As a simple benchmark, capital is allocated equally across all assets.

For \(N\) assets:

$$
w_i = \frac{1}{N}
$$

The portfolio weights therefore sum to one.

```python
n_assets = len(tickers)

equal_weights = np.ones(n_assets) / n_assets
print(equal_weights)
equal_weights.sum()
```

## 3. Expected Return

The expected portfolio return is:

$$
E[R_p] = w^\top \mu
$$

```python
baseline_return = equal_weights @ mu

print(f"Expected annual return: {baseline_return:.2%}")
```

## 4. Portfolio Volatility

Portfolio variance is calculated as:

$$
\sigma_p^2 = w^\top \Sigma w
$$

Annual portfolio volatility is the square root of portfolio variance.

```python
baseline_variance = equal_weights @ cov_matrix @ equal_weights

baseline_volatility = np.sqrt(baseline_variance)

print(f"Annual volatility: {baseline_volatility:.2%}")
```

## 5. Sharpe Ratio

The Sharpe ratio measures excess expected return per unit of portfolio risk:

$$
\text{Sharpe Ratio}
=
\frac{E[R_p]-R_f}{\sigma_p}
$$

For this initial benchmark, the risk-free rate is assumed to be zero. This simplifying assumption will be revisited in later analysis.

```python
risk_free_rate = 0.0

baseline_sharpe = (
    baseline_return - risk_free_rate
) / baseline_volatility

print(f"Sharpe ratio: {baseline_sharpe:.2f}")
```

```python
baseline_summary = pd.DataFrame({
    "Metric": [
        "Expected Annual Return",
        "Annual Volatility",
        "Sharpe Ratio"
    ],
    "Value": [
        baseline_return,
        baseline_volatility,
        baseline_sharpe
    ]
})

baseline_summary
```

The equal-weight portfolio provides a simple, non-optimized benchmark.

Its performance will later be compared with portfolios generated through Monte Carlo simulation and numerical optimization. This allows us to evaluate whether optimization actually improves the portfolio's risk-return trade-off.
